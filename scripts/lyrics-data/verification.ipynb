```python

def show_df_for_year(total_df, year: int):
    df = total_df.filter(pl.col("year") == year)
    display(df)


```

```python
with pl.Config(tbl_rows=25, tbl_cols=-1):
    show_df_for_year(df, 1970)
```

```python
# all songs with multiple artists
with pl.Config(tbl_rows=50, tbl_cols=-1):
    display(df.filter(pl.col("artists").list.len() > 1))
```

```python

# print all songs with queen as an artist
with pl.Config(tbl_rows=-1, tbl_cols=-1):
    display(df.filter(pl.col("artists").list.contains("Queen")))
```

```python
# see if song with pressure is in the data
with pl.Config(tbl_rows=-1, tbl_cols=-1):
    display(df.filter(pl.col("artists").list.contains("TLC")))
```

```python
# print all songs without wikipedia link
with pl.Config(tbl_rows=-1, tbl_cols=-1):
    display(df.filter(pl.col("wikipedia_link").is_null()))
```
